## Summary

* Random signals are signals whose values cannot (or only to a limited extent) be predicted. A random process generates an ensemble of random signals; one particular random signal is termed a sample function. It is not meaningful to describe a random signal by the amplitude values of one sample function. Instead, random signals are characterised by their statistical properties, expressed in terms of amplitude distributions and ensemble averages.

* The cumulative distribution function (CDF) and the probability density function (PDF) characterise the statistics of the amplitude values of a random signal. The bivariate PDF of two random signals or of one random signal at two time instants characterises their statistical dependencies. The PDF can be estimated by the normalised histogram. Analytic distributions, such as the uniform, normal, and Laplace distributions, are used as models for practical random processes. They are described by a few parameters, for instance, the linear mean and variance.

* Ensemble averages are computed across all sample functions for a given time instant $k$. Important first-order ensemble averages are the linear mean $\mu_x[k]$, the quadratic mean $E\{x^2[k]\}$ and the variance $\sigma_x^2[k]$ which are linked by $E\{x^2[k]\} = \mu_x^2[k] + \sigma_x^2[k]$. Second-order ensemble averages consider two time instants and/or two random signals. The linear mapping results in the auto-correlation function (ACF) and the cross-correlation function (CCF). In practice, only a finite number of sample functions is available. The resulting estimates (empirical ensemble averages) are subject to statistical uncertainties that typically decrease with increasing numbers of sample functions.

* A random process is stationary if its statistical properties do not depend on the time index $k$. For wide-sense stationary (WSS) processes, this holds only for the mean and the ACF/CCF. The correlation functions of WSS signals depend only on the difference $\kappa$ between the two time instants. A stationary random process is ergodic if the temporal averages of all sample functions are equal to the ensemble averages. The statistical properties of an ergodic process can then be estimated from a single sample function via temporal averaging. This is the basis for most practical computations of the statistical properties of random signals.

* The ACF $\varphi_{xx}[\kappa]$ of a WSS random signal quantifies the similarity of the signal with a shifted version of itself. It has even symmetry and its maximum at $\kappa = 0$, where $\varphi_{xx}[0] = \mu_x^2 + \sigma_x^2$ is the quadratic mean. For typical random signals, it approaches $\mu_x^2$ for large positve/negative lags $\kappa$. The ACF of a periodic signal is periodic with the same period, which can be exploited, e.g. for pitch estimation. The CCF $\varphi_{xy}[\kappa]$ of two signals can be used, e.g. for the estimation of time-of-arrival from the location of its maximum. For random signals of finite length, the correlation functions can be computed by convolution with a time-reversed signal. A random signal is termed uncorrelated if $\varphi_{xx}[\kappa] = \mu_x^2 + \sigma_x^2 \cdot \delta[\kappa]$. Two random signals are uncorrelated if $\varphi_{xy}[\kappa] = \mu_x \cdot \mu_y$.

* The (auto-) power spectral density (PSD) $\Phi_{xx}(\mathrm{e}^{\,\mathrm{j}\,\Omega})$ is the discrete-time Fourier transform (DTFT) of the ACF. It is real-valued, has even symmetry and quantifies the average power per frequency of a random signal. The integral over the PSD (normalised by $2 \pi$) yields the quadratic mean $\varphi_{xx}[0]$. Analogously, the cross-power spectral density is the DTFT of the CCF.

* Two random signals are independent if their bivariate PDF is equal to the product of their univariate PDFs. Independent random signals are also uncorrelated. They are orthogonal if at least one of them is zero-mean. The reverse, that uncorrelated or orthogonal signals are independent, does not hold in general.

* White noise is a WSS random signal with constant PSD $\Phi_{xx}(\mathrm{e}^{\,\mathrm{j}\,\Omega}) = N_0$, its ACF is given as $\varphi_{xx}[\kappa] = N_0 \cdot \delta[\kappa]$. White noise is zero-mean, and its samples are uncorrelated. The amplitude distribution of white noise is not necessarily normal; this must be stated explicitly.

* The linear mean of a superposition $y[k] = x[k] + n[k]$ of two random signals is given by the sum of the linear means. The ACF of the superposition is the sum of all auto- and cross-correlation functions of both signals. For two uncorrelated signals, at least one of which is zero-mean, the ACFs and PSDs of both signals simply add up. The PDF of the superposition of two independent signals is given by the convolution of their PDFs. According to the central limit theorem, the PDF of a superposition of many independent random signals tends towards a normal distribution. The superposition of a signal with uncorrelated zero-mean white noise drawn from a normal distribution is known as the additive white Gaussian noise (AWGN) model. Averaging over $N$ observations disturbed by mutually independent noise improves the signal-to-noise ratio by $10 \cdot \log_{10}(N)$ dB.

**Interactive Example**

The following interactive example illustrates the statistical properties of a superposition of an uncorrelated random signal $x[k]$ and a cosine signal with random phase $\varphi$

\begin{equation}
y[k] = x[k] + A \cdot \cos(\Omega_0 k + \varphi)
\end{equation}

where $A \in \mathbb{R}$ denotes the amplitude and $\Omega_0$ the normalized frequency of the cosine signal. The amplitude distribution of $x[k]$ can be chosen as uniform, normal or Laplace distribution with linear mean $\mu_x$ and variance $\sigma_x^2$. For $\mu_x = 0$ the random signal $x[k]$ is white noise. The phase $\varphi$ is drawn from a uniform distribution in $[0, 2 \pi)$ for each sample function and is constant within the sample function. Due to the random phase, the cosine signal is a stationary random process with zero linear mean, whose ACF depends only on the lag $\kappa$. For a fixed phase, e.g. $\varphi = 0$ for all sample functions, the linear mean $E\{ y[k] \} = \mu_x + A \cos(\Omega_0 k)$ would depend on $k$ and the superposition would not be stationary. It is assumed that the random signal and the cosine signal are independent. The ACF of the superposition is then given as

\begin{equation}
\varphi_{yy}[\kappa] = \mu_x^2 + \sigma_x^2 \cdot \delta[\kappa] + \frac{A^2}{2} \cos(\Omega_0 \kappa)
\end{equation}

and its PSD as

\begin{equation}
\Phi_{yy}(\mathrm{e}^{\,\mathrm{j}\,\Omega}) = \sigma_x^2 + \mu_x^2 \cdot {\bot \!\! \bot \!\! \bot}\left( \frac{\Omega}{2 \pi} \right) + \frac{A^2}{4} \left( {\bot \!\! \bot \!\! \bot}\left( \frac{\Omega - \Omega_0}{2 \pi} \right) + {\bot \!\! \bot \!\! \bot}\left( \frac{\Omega + \Omega_0}{2 \pi} \right) \right).
\end{equation}

The PDF $p_y(\theta)$ of the superposition is given by the convolution of the PDF of $x[k]$ with the PDF of the cosine signal (arcsine distribution).

One sample function of length $N$ is generated. Its first samples are plotted together with the linear mean $\overline{y[k]}$ (dashed line) and the range $\overline{y[k]} \pm \hat{\sigma}_y$ (shaded area), where the standard deviation $\hat{\sigma}_y = \sqrt{\overline{y^2[k]} - \overline{y[k]}^2}$ is computed from the linear and quadratic mean estimated by temporal averaging. Theory yields $\mu_y = \mu_x$, $E\{ y^2[k] \} = \varphi_{yy}[0] = \mu_x^2 + \sigma_x^2 + \frac{A^2}{2}$ and $\sigma_y^2 = \sigma_x^2 + \frac{A^2}{2}$. The histogram $\hat{p}_y(\theta)$ is compared to the PDF $p_y(\theta)$, the estimated ACF $\hat{\varphi}_{yy}[\kappa]$ to the ACF $\varphi_{yy}[\kappa]$. The PSD $\hat{\Phi}_{yy}(\mathrm{e}^{\,\mathrm{j}\,\Omega})$ is estimated by the DFT of the estimated ACF. The dashed line indicates the constant level $\sigma_x^2$ of the PSD of the random signal $x[k]$. The Dirac impulses of the PSD at $\Omega = 0$ (linear mean $\mu_x$) and at $\Omega = \pm \Omega_0$ (cosine signal) show up as distinct peaks in the estimated PSD. Note that the estimated PSD shows strong fluctuations which do not decrease with an increasing number $N$ of samples. Methods to estimate the PSD more reliably are discussed in the chapter on spectral estimation. You can adjust the parameters of the random signal, the cosine signal and the number of samples interactively.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

from ipywidgets import interactive_output, FloatSlider, IntSlider, Dropdown
from ipywidgets import HBox, VBox, Label, Layout
from IPython.display import display


def distribution(name, mu, sigma):
    """Returns frozen distribution with linear mean mu and standard deviation sigma."""
    if name == "uniform":
        return stats.uniform(loc=mu - np.sqrt(3) * sigma, scale=2 * np.sqrt(3) * sigma)
    elif name == "normal":
        return stats.norm(loc=mu, scale=sigma)
    elif name == "Laplace":
        return stats.laplace(loc=mu, scale=sigma / np.sqrt(2))


def plot_random_signal(dist, mu, sigma, A, Om0, N):
    # superposition of random signal and cosine signal with random phase
    # (the phase is drawn once and is constant within the sample function)
    rng = np.random.default_rng(seed=1)
    px = distribution(dist, mu, sigma)
    x = px.rvs(size=N, random_state=rng)
    phi = rng.uniform(0, 2 * np.pi)
    k = np.arange(N)
    y = x + A * np.cos(Om0 * k + phi)

    # theoretical PDF by convolution of PDFs (average over phase of cosine)
    theta = np.linspace(-6, 6, 1201)
    ph = np.linspace(0, 2 * np.pi, 1000, endpoint=False)
    py = np.mean(px.pdf(theta[:, np.newaxis] - A * np.cos(ph)), axis=1)

    # estimated (biased) and theoretical ACF for lags -K <= kappa <= K
    K = 32
    kappa = np.arange(-K, K + 1)
    acf = 1 / N * np.correlate(y, y, mode="full")[N - 1 - K : N + K]
    acf_th = mu**2 + sigma**2 * (kappa == 0) + A**2 / 2 * np.cos(Om0 * kappa)

    # estimated PSD by DFT of estimated ACF, evaluated for -pi <= Omega < pi
    Om = np.fft.fftshift(np.fft.fftfreq(N)) * 2 * np.pi
    psd = np.fft.fftshift(1 / N * np.abs(np.fft.fft(y)) ** 2)
    eps = np.nextafter(0, 1)

    fig, ax = plt.subplots(2, 2, figsize=(10, 8))
    # sample function with temporal averages
    mean = np.mean(y)
    qmean = np.mean(y**2)
    std = np.sqrt(qmean - mean**2)
    ax[0, 0].axhspan(
        mean - std,
        mean + std,
        color="C1",
        alpha=0.2,
        label=r"$\overline{y[k]} \pm \hat{\sigma}_y$",
    )
    ax[0, 0].axhline(mean, color="C1", linestyle="--", label=r"$\overline{y[k]}$")
    ax[0, 0].plot(k[:100], y[:100], color="C0")
    ax[0, 0].set_xlabel(r"$k$")
    ax[0, 0].set_ylabel(r"$y[k]$")
    ax[0, 0].set_title(
        r"$\overline{{y[k]}}$ = {:.2f}, $\overline{{y^2[k]}}$ = {:.2f}, $\hat{{\sigma}}_y^2$ = {:.2f}".format(
            mean, qmean, std**2
        )
    )
    ax[0, 0].set_xlim([0, 99])
    ax[0, 0].set_ylim([-6, 6])
    ax[0, 0].legend(loc="upper right")
    ax[0, 0].grid()
    # PDF
    ax[0, 1].hist(y, bins=100, range=(-6, 6), density=True, color="C1", alpha=0.5)
    ax[0, 1].plot(theta, py, label=r"$p_y(\theta)$")
    ax[0, 1].set_xlabel(r"$\theta$")
    ax[0, 1].set_ylabel(r"$\hat{p}_y(\theta)$, $p_y(\theta)$")
    ax[0, 1].set_xlim([-6, 6])
    ax[0, 1].set_ylim(bottom=0)
    ax[0, 1].legend(loc="upper right")
    ax[0, 1].grid()
    # ACF
    ax[1, 0].stem(
        kappa,
        acf,
        label=r"$\hat{\varphi}_{yy}[\kappa]$",
        basefmt=" ",
        linefmt="C1",
        markerfmt="C1o",
    )
    ax[1, 0].plot(kappa, acf_th, "C0x", label=r"$\varphi_{yy}[\kappa]$")
    ax[1, 0].set_xlabel(r"$\kappa$")
    ax[1, 0].set_ylabel(r"$\hat{\varphi}_{yy}[\kappa]$, $\varphi_{yy}[\kappa]$")
    ax[1, 0].set_xlim([-K, K])
    ax[1, 0].set_ylim([-2.5, 4.5])
    ax[1, 0].legend(loc="upper right")
    ax[1, 0].grid()
    # PSD
    ax[1, 1].plot(
        Om,
        10 * np.log10(psd + eps),
        color="C1",
        label=r"$\hat{\Phi}_{yy}(e^{j\Omega})$",
    )
    ax[1, 1].axhline(
        10 * np.log10(sigma**2), color="C0", linestyle="--", label=r"$\sigma_x^2$"
    )
    ax[1, 1].set_xlabel(r"$\Omega$")
    ax[1, 1].set_ylabel(r"$\hat{\Phi}_{yy}(e^{j\Omega})$ in dB")
    ax[1, 1].set_xlim([-np.pi, np.pi])
    ax[1, 1].set_ylim([-50, 50])
    ax[1, 1].legend(loc="upper right")
    ax[1, 1].grid()

    fig.tight_layout()
    plt.show()


# widgets for random signal, cosine signal and estimation
controls = dict(
    dist=Dropdown(
        options=["uniform", "normal", "Laplace"], value="normal", description="PDF"
    ),
    mu=FloatSlider(min=-1, max=1, step=0.05, value=0, description="μₓ"),
    sigma=FloatSlider(min=0.1, max=1.5, step=0.05, value=0.5, description="σₓ"),
    A=FloatSlider(min=0, max=2, step=0.05, value=1, description="A"),
    Om0=FloatSlider(min=0, max=np.pi, step=0.01, value=0.5, description="Ω₀"),
    N=IntSlider(min=128, max=16384, step=128, value=4096, description="N"),
)


def heading(text):
    return Label(
        text, style=dict(font_weight="bold"), layout=Layout(align_self="center")
    )


ui = HBox(
    [
        VBox(
            [
                heading("random signal"),
                controls["dist"],
                controls["mu"],
                controls["sigma"],
            ]
        ),
        VBox([heading("cosine signal"), controls["A"], controls["Om0"]]),
        VBox([heading("number of samples"), controls["N"]]),
    ]
)
display(ui, interactive_output(plot_random_signal, controls))
```
